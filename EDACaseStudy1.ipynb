```python
#outlier -> extremely low and extremely high
#boxplot
#apply a loop and plot the boxplot for all the columns
for i in df.columns:
  if df[i].dtype=="float64":
    plt.boxplot(df[i])
    plt.xlabel(i)
    plt.show()
```

```python
#Removing the outliers
Q1 = df.oil_brent.quantile(0.25)
Q3 = df.oil_brent.quantile(0.75)
IQR = Q3-Q1
LL = Q1 - 1.5*IQR
UL = Q3 + 1.5*IQR
df = df[(df["oil_brent"]>=LL)&(df['oil_brent']<=UL)]
plt.boxplot(df["oil_brent"])
```

```python
for i in df.columns:
  if df[i].dtype=="float64":
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3-Q1
    LL = Q1 - 1.5*IQR
    UL = Q3 + 1.5*IQR
    df = df[(df[i]>=LL)&(df[i]<=UL)]
```

```python
for i in df.columns:
  if df[i].dtype=="float64":
    plt.boxplot(df[i])
    plt.xlabel(i)
    plt.show()
```
